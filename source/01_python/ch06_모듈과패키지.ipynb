# 1절. 파이썬 표준 모듈 사용

```
파이썬 설치하는 방법
    - python.org/microsoft store : 파이썬 표준모듈200여개가 포함된 파이썬 인터프리터를 install(sys, re, datetime, time...)
    - miniconda : python.org버전 보다 조금 많은 모듈이 포함된 인터프리터를 install
    - anaconda : 몇백개 이상의 패키지, 개발도구(IDE tools;jupyter notebook)를 포함한 인터프리터를 설치(200명이상의 기업체에서는 유료)
    - uv(pip-파이썬 패키지 설치도구, 패키지/프로젝트 관리도구)를 이용한 설치
        * 참조 : https://docs.astral.sh/uv/getting-started/installation/
        * uv설치 후 파이썬 설치(반드시 powershell관리자권한으로 실행)
        
        * Set-ExecutionPolicy RemoteSigned -Scope CurrentUser 
        * Get-ExecutionPolicy명령으로 RemoteSigned출력확인
        * powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex" 
        * powershell 닫았다 다시 열기
        * uv python list
        * uv python install 3.11(업계에서는 3.13이상은 잘 안 씀) - 지금은 하지 마셔요
```

````python
# 1. import 모듈명 : 모듈안의 함수, 클래스, 변수를 모듈명.함수명()으로 사용
import time # time모듈안의 함수, 변수, 클래스를 메모리 로드
time.ctime() # 현재 날짜와 시간
````

````python
time.time()  # 1970.1.1부터의 초
````

````python
%%time
import time
start = time.time() 
cnt = 0
while cnt<=10000000:
    cnt += 1
end = time.time()
print('반복문 1천만번 돈 시간 :', (end-start), '초')
````

````python
del time # time모듈을 메모리에서 unload
````
