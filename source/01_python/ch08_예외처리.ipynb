# 2절. try ~ except로 예외처리
## 2.1 try~except
```
try:
    예외가 발생할 가능성이 있는 명령어들
except:
    예외가 발생했을 경우 실행할 명령어들
```

````python
# 100을 입력받은 정수값으로 나눠 출력
x = int(input('정수를 입력하세요 >'))
print('입력한 정수는 {}'.format(x))
print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
# ZeroDivisionError : "division by zero" 
#                     0을 입력했을 때 예외타입
# ValueError : invalid literal for int() with base 10: 'a'
#              invalid literal(잘못된 문자)를 입력할 경우
print('DONE')
````

````python
# 100을 입력받은 정수값으로 나눠 출력
# ZeroDivisionError : "division by zero" 
# ValueError : invalid literal for int() with base 10: 'a'
try:
    x = int(input('정수를 입력하세요 >'))
    print('입력한 정수는 {}'.format(x))
    print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
except:
    print('나누기에 유효한 정수가 아닙니다')
print('DONE')
````

````python
# 100을 나눌 유효한 정수를 입력할 때까지 정수를 입력받아 유효한 정수를 입력하면 입력한 정수와 100으로 나눈 수를 출력
# ZeroDivisionError : "division by zero" 
# ValueError : invalid literal for int() with base 10: 'a'
while True:
    try:
        x = int(input('정수를 입력하세요 >'))
        print('입력한 정수는 {}'.format(x))
        print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
        break
    except:
        print('나누기에 유효한 정수가 아닙니다')
print('DONE')
````

## 2.2 예외를 지정한 처리
```
try:
    예외가 발생할 수도 있는 문장들
except 예외타입1:
    해당 예외가 발생할 경우 실행할 문장
except 예외타입2:
    해당 예외가 발생할 경우 실행할 문장
```
**여러 예외처리를 할 경우 상위 예외타입은 아래로**

````python
# 100을 나눌 유효한 정수를 입력할 때까지 정수를 입력받아 유효한 정수를 입력하면 입력한 정수와 100으로 나눈 수를 출력
# ZeroDivisionError : "division by zero" 
# ValueError : invalid literal for int() with base 10: 'a'
while True:
    try:
        x = int(input('100을 나눌 정수를 입력하세요 >'))
        print('입력한 정수는 {}'.format(x))
        print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
        break    
    except ValueError:
        print('정수가 아닙니다. 다시 입력하세요')
    except ZeroDivisionError:
        print('0으로 나눌 수 없습니다. 다시 입력하세요')    
    except Exception:
        print('원인 모를 예외가 발생했습니다. 다시 입력하세요')
print('DONE')
````

````python
# 100을 나눌 유효한 정수를 입력할 때까지 정수를 입력받아 유효한 정수를 입력하면 입력한 정수와 100으로 나눈 수를 출력
# ZeroDivisionError : "division by zero" 
# ValueError : invalid literal for int() with base 10: 'a'
while True:
    try:
        x = int(input('100을 나눌 정수를 입력하세요 >'))
        print('입력한 정수는 {}'.format(x))
        print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
        break    
    except (ValueError, ZeroDivisionError):
        print('100으로 나눌 수 있는 정수가 아닙니다. 다시 입력하세요')   
    except Exception:
        print('원인 모를 예외가 발생했습니다. 다시 입력하세요')
print('DONE')
````

## 2.3 예외인수(e)
```
try:
    예외가 발생할 수도 있는 문장들
except 예외타입 as e:
    print(e.args[0])
    print(e)
```

````python
e = ZeroDivisionError("division by zero") # 예외가 발생되면 자동적으로 예외객체가 생성
e.args[0]
````

````python
while True:
    try:
        x = int(input('100을 나눌 정수를 입력하세요 >'))
        print('입력한 정수는 {}'.format(x))
        print('100을 입력한 정수로 나누면 {:.2f}'.format(100/x))
        break    
    except (ValueError, ZeroDivisionError) as e:
        print(e)   
        # print(e.args[0])
    except Exception:
        print('원인 모를 예외가 발생했습니다. 다시 입력하세요')
print('DONE')
````

```
try:
    명령어들
except [예외타입 [as e]]:
    해당 예외타입시 수행할 명령어
[else:
    try블록을 다 수행한 후 실행할 명령어 ]
finally:
    try블록이나 except블록을 수행한 후 실행
```
