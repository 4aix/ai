# 1절. txt파일에 데이터 저장하고 불러오기
## 1.1 write

```python
f = open('data/ch09.txt', 'w') # encoding='cp949'(win) / 'utf-8'(mac, linux)
# w 또는 wt : 기존의 파일이 있는 경우 덮어쓰기, 파일이 없는 경우 생성하여 쓰기
# a 또는 at : 기존의 파일이 있는 경우 append, 파일이 없는 경우 생성하고 쓰기
# r 또는 rt : mode의 기본값. 파일이 있는 경우 파일을 열고, 파일이 없는 경우 예외 발생
print("쓰기 가능한지 :", f.writable())
```

```python
f.write('Hello\nWorld\n') # 한글을 write하면 cp949로 저장(win)
f.close()
```

```python
# with 구문을 사용하면 매번 close하는 코드가 불필요
with open('data/ch09.txt', 'a') as f: # encoding='cp949'
    f.write('Python is easy.\nPython is smart.\n')
```

```python
with open('data/ch09.txt', 'a') as f: # encoding='cp949'
    print('print함수는 파일출력, 화면출력\n', file=f)
# 주피터노트북에서는 utf-8 인코딩방식으로 한글text가 열려.
```

- utf-8 / UTF-8 / UTF8 : 한글조합형 - 주피터노트북에서 
- euc-kf : 한글완성형 (믜x)
- cp949 : 확장된 한글완성형(믜O) - open()함수 기본 encoding방식, 엑셀에서

```python
with open('data/ch09.txt', 'w', encoding='utf-8') as f:
    # 1방법
    f.write('홍길동,30,서울\n')
    f.write('김길동,25,안양\n')
    # 2방법
    text_list = ['홍길동,30,서울\n', '김길동,25,안양\n']
    for line in text_list:
        f.write(line)
    # 3방법
    f.writelines(text_list)
```

## 1.2 read

```python
# 1방법 : 파일을 한번에 읽어오기
with open('data/ch09.txt', 'r', encoding='utf-8') as txtfile:
    filecontent = txtfile.read()
filecontent
```

```python
# 2방법 : 파일을 한줄씩 읽고 출력
with open('data/ch09.txt', 'rt', encoding='utf-8') as txtfile:
    line = txtfile.readline() # 한줄 읽기
    while line!= '' :
        print(line, end='') # line에 \n이 포함되어 있음
        line = txtfile.readline()
```

```python
# 3방법 : 파일을 한꺼번에 여러줄 읽어 list로 받기
with open('data/ch09.txt', 'r', encoding='UTF-8') as txtfile:
    lines = txtfile.readlines()
print(lines)
```
