```python
person1 = {'name':'홍길동', 'gender':'남자', 'academy':'MBC'}
person2 = {'name':'홍길숙', 'gender':'여자', 'academy':'MBC'}
print(person1.get('name'), person1['gender'], person1.get('academy', '-'))
print(person1)
print(person2)
```

```python
class Person:
    academy = 'MBC' # 클래스 변수
    def __init__(self, name, gender):# self:내 객체 자신
        self.name = name
        self.gender = gender
        print('Person 객체 생성합니다')
    def print_info(self): 
        print("{}의 {}은/는 {}입니다".format(self.academy,
                                      self.name,
                                      self.gender))
p1 = Person('홍길동', '남자')
print(p1)
p1.print_info()
p1.name = '김길동'
p1.academy = '딴데'
p1.print_info()
```

```python
p2 = Person('김길순', '여자')
p2.print_info()
p2.name="김낄쑥"
p2.print_info()
```

```python
p1.academy, p2.academy, Person.academy
```
