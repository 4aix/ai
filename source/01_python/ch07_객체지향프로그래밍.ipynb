````python
person1 = {'name':'홍길동', 'gender':'남자', 'academy':'MBC'}
person2 = {'name':'홍길숙', 'gender':'여자', 'academy':'MBC'}
print(person1.get('name'), person1['gender'], person1.get('academy', '-'))
print(person1)
print(person2)
````

````python
class Person:
    academy = 'MBC' # 클래스 변수
    def __init__(self, name, gender):# self:내 객체 자신
        self.name = name
        self.gender = gender
        print('Person 객체 생성합니다')
    def print_info(self): 
        print("{}의 {}은/는 {}입니다".format(self.academy,
                                      self.name,
                                      self.gender))
p1 = Person('홍길동', '남자')
print(p1)
p1.print_info()
p1.name = '김길동'
p1.academy = '딴데'
p1.print_info()
````

````python
p2 = Person('김길순', '여자')
p2.print_info()
p2.name="김낄쑥"
p2.print_info()
````

# 2절. 생성자(`__init__()`)와 소멸자(`__del__()`)
```
생성자 : __init__(self, 매개변수, ..) 객체가 생성될 때 자동 실행
        생성시 필요한 코드(인스턴스 변수 초기화)를 포함
        없으면 default 생성자가 자동 생성
소멸자 : __del__(self) 객체가 소멸될(참조한 변수가 없어질) 때 자동 실행
        인스턴스 객체의 레퍼런스 카운트가 0일 될 때 실행
        소멸시 필요한 코드를 포함
```

````python
class Person:
    def __init__(self, name, gender):
        self.name = name     # 인스턴스 변수
        self.gender = gender
        print('Person 생성자 호출(객체 생성 == 클래스형 변수 생성 == 인스턴스 메모리 생성)')
    def __del__(self):
        print('Person 객체 소멸')
    def print_info(self):
        print(f"{self.name}님은 {self.gender}입니다")
    def __str__(self):
        return f"{self.name}은 {self.gender}이죠"
````

````python
p1 = Person('신사임당', '여자')
p1.print_info()
print(p1.__str__())
print(p1)
````
